```python
def count_binary(n): 
        for num in range(n + 1):
            binary = ""   # This string will store the binary representation of the current number as it's built digit by digit.
            current = num
            
            # Construct the binary representation
            while current > 0:
                binary = str(current % 2) + binary  # Get remainder and prepend
                current //= 2  # Divide by 2
                
            if binary == "":  # Case for 0
                    binary = "0"
                
            print(binary)

n = int(input("Enter an positive integer: "))

# Check if the number is positive
if n < 0:
    print(f"{n} is not a positive integer.")
else:
    count_binary(n)
            
```

```python
#Alt solution to task 1
def count_binary(n): 
    for num in range(n + 1):
        # Using the bin() function to convert the number to binary
        binary = bin(num)[2:]  # bin() returns a string starting with '0b' so slice off the first two characters
        
        print(binary)

n = int(input("Enter a positive integer: "))

# Check if the number is positive
if n < 0:
    print(f"{n} is not a positive integer.")
else:
    count_binary(n)
```

**Expected Output**

```python
# Define the Vector class
class Vector:
    # Initialize the vector with x and y components
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overload the '+' operator for adding two vectors
    def __add__(self, other):
        # Add the x components and y components separately
        new_x = self.x + other.x
        new_y = self.y + other.y
        # Return a new Vector object with the resulting components
        return Vector(new_x, new_y)

    # Provide a string representation of the vector
    def __str__(self):
        # Format the vector as (x, y)
        return f"({self.x}, {self.y})"


# Example usage of the Vector class
if __name__ == "__main__":
    # Create two vector objects
    v1 = Vector(1, 2)
    v2 = Vector(3, 4)

    # Add the vectors using the overloaded '+' operator
    v3 = v1 + v2

    # Print the resulting vector using the overloaded 'str' method
    print("Sample Output 1:")
    print(v3)  # Output: (4, 6)

    # Test with different vector values
    print("\nSample Output 2:")
    v4 = Vector(-2, 5)
    v5 = Vector(4, -3)
    v6 = v4 + v5
    print(v6)  # Output: (2, 2)
```

**Explanation to sample outputs:**  
* Sample Output 1:  
Adds v1 = Vector(1, 2) and v2 = Vector(3, 4).  
The result is v3 = Vector(4, 6).    
* Sample Output 2:  
Adds v4 = Vector(-2, 5) and v5 = Vector(4, -3).

```python
import random
import time

# 1. Generate random floats
def generate_random_floats(count, lower, upper):
    # Generate a list of random float numbers within a specified range.
    return [random.uniform(lower, upper) for _ in range(count)]

# 2 & 3. Analyze behavior of int() & round()
def int_and_round(random_floats):
    print(f"{'Original':<10} {'int()':<10} {'round()':<10}")
    for num in random_floats:
        print(f"{num:<10.2f} {int(num):<10} {round(num):<10}")

# 4. Analyze Behavior with 0.5
def test_case():
    test_values = [2.5, 3.5, -2.5, -3.5]
    print("\nTesting behavior with 0.5:")
    print(f"{'Original':<10} {'int()':<10} {'round()':<10}")
    for num in test_values:
        print(f"{num:<10} {int(num):<10} {round(num):<10}")

# 5. Compare Performance
def compare_performance(random_floats):
    start_time = time.time()
    [int(num) for num in random_floats]
    int_time = time.time() - start_time

    start_time = time.time()
    [round(num) for num in random_floats]
    round_time = time.time() - start_time

    print("\nPerformance Comparison:")
    print(f"int() time: {int_time:.6f} seconds")
    print(f"round() time: {round_time:.6f} seconds")

# Main Execution
random_floats = generate_random_floats(10, -10, 10)  # Generate 10 random floats
int_and_round(random_floats)
test_case()

large_dataset = generate_random_floats(1_000_000, -10, 10)  # Large dataset for performance comparison
compare_performance(large_dataset)
```

## Task 5

**Algorithm Steps** 

**1.** Get input range to extract min_area and max_area  
**2.** Initialize an empty dictionary to store the results  
**3.** Loop through the areas in the range  
**4.** Find all valid (length, width) pairs like: length * width = area  
**5.** Add the area and its combinations to the dictionary  
**6.** Output the dictionary with the garden catalogs

**Expected output:**   
Enter the range (e.g.: 1,16):  **1,16**  
Unique garden catalog of 7: [(1, 7), (7, 1)  ]
Unique garden catalog of 14: [(1, 14), (14, 1), (2, 7), (7, 2)]




```python
def unique_garden_catalog(area_range):
    # 1. Parse the input string to extract min_area and max_area
    min_area, max_area = map(int, area_range.split(','))
    
    # 2. Initialize an empty dictionary to store the results
    garden_catalog = {}
    
    # 3. Iterate through the areas in the range
    for area in range(min_area, max_area + 1):
        if area % 7 == 0:  # Check if the area is divisible by 7
            combinations = []  # List to store (length, width) combinations
            
            # 4. Find all valid (length, width) pairs: length * width = area
            for length in range(1, int(area**0.5) + 1):
                if area % length == 0:  # If length divides the area
                    width = area // length  # Compute width
                    # Add both (length, width) and (width, length) as unique combinations
                    combinations.append((length, width))
                    if length != width:  # Avoid duplicates for square numbers
                        combinations.append((width, length))
            
            # Step 5: Add the area and its combinations to the dictionary
            garden_catalog[area] = combinations
    
    # 6. Output the dictionary with the garden catalogs
    for area, pairs in garden_catalog.items():
        print(f"Unique garden catalog of {area}: {pairs}")

user_input = input("Enter the range (e.g.: 1,16): ")
unique_garden_catalog(user_input)
```

**2.** jupyter notebook  
* The notebook imports the get_factorial module.

**Expected output:**  
The factorial of 5 is 120

Error calculating factorial of -2: Input must be a non-negative integer.

```python
import get_factorial

# List of test cases
numbers = [5, 0, 10, -2, 3.5]

# Loop for calculating the factorial of the numbers
for num in numbers:
    result = get_factorial.factorial(num)
    if isinstance(result, int):  # If result is an integer (valid factorial)
        print(f"The factorial of {num} is {result}")
    else:
        print(f"Error calculating factorial of {num}: {result}")
```
